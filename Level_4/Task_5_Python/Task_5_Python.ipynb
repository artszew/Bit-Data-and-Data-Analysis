1. Suma liczby zachorowań ze względu na kraj

```python
from functools import reduce
import csv
```

```python

with open("../final_data.csv", newline='', encoding='utf-8') as f:
    reader = csv.reader(f)
    next(reader)

    #Wybranie wszystkich wierszy gdzie kolumna Total_confirmed nie ma pustych wartości np. ""
    filtered = filter(lambda row: row[6].isdigit(), reader)

    # Map
    mapped = map(lambda row: (row[0], int(row[6])), filtered)

    # Reduce
    def reducer(acc, item):
        country, cases = item
        acc[country] = acc.get(country, 0) + cases
        return acc

    result = reduce(reducer, mapped, {})



for country, total_cases in sorted(result.items()):
    print(f"{country} {total_cases}")
```

2. Udział osób w pełni zaszczepionych w populacji dla kraju

```python
with open("../final_data.csv", newline='', encoding='utf-8') as f:
    reader = csv.reader(f)
    next(reader)

    # Poprawne wiersze z danymi o szczepieniach i populacji
    filtered = filter(lambda row: row[9].isdigit() and row[20].isdigit(), reader)

    # Map
    mapped = map(lambda row: (row[0], (int(row[9]), int(row[20]))), filtered)

    # Reduce
    def reducer(acc, item):
        country, (vaccinated, population) = item
        if country in acc:
            acc[country]['vaccinated'] += vaccinated
        else:
            acc[country] = {'vaccinated': vaccinated, 'population': population}
        return acc

    result = reduce(reducer, mapped, {})


for country, data in sorted(result.items()):
    vaccinated = data['vaccinated']
    population = data['population']
    percentage = (vaccinated / population) * 100 if population else 0
    print(f"{country}\t{percentage:.2f}%")
```

1. Wprowadzenie:
- map(f, kolekcja): przekształca każdy element
- filter(f, kolekcja): wybiera elementy spełniające warunek
- reduce(f, kolekcja): agreguje dane do jednej wartości

2. Przykład (COVID-19):
Dane: ["Polska,2021-01-01,100", "Polska,2021-01-02,200", ...]
- map: podział danych po przecinku
- filter: wybór rekordów dla Polski
- reduce: sumowanie przypadków

3. Porównanie z Hadoop MapReduce:

| Cecha           | Python map/filter/reduce | Hadoop MapReduce     |
|----------------|---------------------------|----------------------|
| Skalowalność   | Tylko lokalnie            | Rozproszone (klaster)|
| Wydajność      | Małe dane                 | Duże dane            |
| Prosta nauka   | Tak                       | Trudniej             |
| Zastosowanie   | Skrypty, analizy          | Big Data             |

4. Wnioski:
- Python: szybki i prosty do małych danych
- Hadoop: lepszy dla bardzo dużych danych i przetwarzania rozproszonego
